### Завдання 28
Для матриці:

```text

     [2, 6]]
```

обчисліть:

- обернену матрицю.

Перевірте результат, помноживши початкову матрицю на обернену.

````python
A = np.array([[4, 7], [2, 6]])
inv = np.linalg.inv(A)
print('Визначник:', np.linalg.det(A))
print('Обернена матриця:', inv, sep='\n')
print('Перевірка:', A @ inv, sep='\n')
print(np.allclose(A @ inv, np.eye(2)))
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` : середнє значення стовпця, `σ` : стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
A = np.array([[10, 20, 30], [20, 30, 40], [30, 40, 50]])
z = (A - A.mean(axis=0)) / A.std(axis=0)
print(z)
print('Середні стовпців:', z.mean(axis=0))
print('Стандартні відхилення:', z.std(axis=0))
````

Для кожного рядка знайдіть його норму та створіть нову матрицю, у якій кожен рядок нормалізований до одиничної довжини.

Спробуйте виконати завдання без циклів.

````python
A = np.arange(1, 17).reshape(4, 4)
norms = np.linalg.norm(A, axis=1, keepdims=True)
normalized = A / norms
print('Норми:', norms.ravel())
print(normalized)
print('Нові норми:', np.linalg.norm(normalized, axis=1))
````

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
A = np.array([[1, 2, 1], [2, 1, -1], [3, -1, 2]])
b = np.array([9, 8, 3])
x = np.linalg.solve(A, b)
print('x, y, z:', x)
print('A @ x:', A @ x)
print('b:', b)
print('Розв’язок правильний:', np.allclose(A @ x, b))
````
